# GitHub File Types — Jupyter Notebook

GitHub renders `.ipynb` files as formatted notebooks — including markdown cells, code cells, and their **outputs**.

This is one of GitHub's most powerful rendering features for data science.

```python
# GitHub renders code cells AND their outputs
languages = {
    'JavaScript': 20, 'Python': 16,
    'TypeScript': 12, 'Java': 8, 'C++': 6
}
print('Top 5 languages by repo count:')
for lang, count in sorted(languages.items(), key=lambda x: -x[1]):
    bar = '█' * (count // 1)
    print(f'{lang:<12}: {bar} {count}')
```

## Key Facts

- GitHub renders markdown cells as formatted HTML
- Code outputs (text, tables, images) are shown inline
- `matplotlib` plots render as embedded images
- Files over ~5MB won't render — use nbviewer.jupyter.org
- Use `nbstripout` to clear outputs before committing
